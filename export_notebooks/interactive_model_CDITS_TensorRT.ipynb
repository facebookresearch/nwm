```python
import tensorrt as trt
import torch
import numpy as np

print("Booting Pure TensorRT Pipeline... (Loading 3 Engines to VRAM)")

# 1. Initialize the TensorRT Runtime
trt_logger = trt.Logger(trt.Logger.WARNING)
trt_runtime = trt.Runtime(trt_logger)

def load_engine(path):
    with open(path, "rb") as f:
        return trt_runtime.deserialize_cuda_engine(f.read())

# 2. Load Engines and Create Contexts
enc_engine = load_engine("./onnx_models/enc.engine")
enc_context = enc_engine.create_execution_context()

cdit_engine = load_engine("./onnx_models/cdit.engine")
cdit_context = cdit_engine.create_execution_context()

dec_engine = load_engine("./onnx_models/dec.engine")
dec_context = dec_engine.create_execution_context()

print("Overriding model_forward_wrapper for PURE BARE-METAL TensorRT...")

def pure_trt_model_forward_wrapper(all_models, curr_obs, curr_delta, num_timesteps, latent_sz, dev, num_cond, num_goals=1, rel_t=None, progress=False):
    _, diff_obj, _ = all_models
    x = curr_obs.to(dev)
    y = curr_delta.to(dev)
    B, T = x.shape[:2]

    if rel_t is None:
        rel_t = (torch.ones(B)* (1. / 128.)).to(dev)
        rel_t *= num_timesteps

    # The dedicated high-speed CUDA stream for all TensorRT operations
    trt_stream = torch.cuda.Stream(device=dev)

    # ---------------------------------------------------------
    # UNIVERSAL TENSORRT EXECUTION HELPER
    # ---------------------------------------------------------
    def run_trt_engine(context, inputs_dict, output_name, out_shape):
        out_tensor = torch.empty(out_shape, dtype=torch.float32, device=dev).contiguous()
        
        # Bind inputs
        for name, tensor in inputs_dict.items():
            context.set_tensor_address(name, tensor.data_ptr())
            
        # Bind output
        context.set_tensor_address(output_name, out_tensor.data_ptr())
        
        # Execute on dedicated stream
        context.execute_async_v3(stream_handle=trt_stream.cuda_stream)
        
        # Tell PyTorch to wait for TRT to finish before moving on
        torch.cuda.current_stream(device=dev).wait_stream(trt_stream)
        return out_tensor

    # ---------------------------------------------------------
    # 1. ENCODE CONTEXT (VAE)
    # ---------------------------------------------------------
    x_flat = x.flatten(0, 1).contiguous().float()
    lat_list = []
    
    # Because your ONNX was exported with Batch=1, we loop through the frames.
    # But because it's pure GPU pointer math now, this loop takes essentially 0.0001s
    for i in range(x_flat.shape[0]):
        px_in = x_flat[i:i+1]
        lat_out = run_trt_engine(enc_context, {"px": px_in}, "lat", (1, 4, latent_sz, latent_sz))
        lat_list.append(lat_out)
        
    x_enc = torch.cat(lat_list, dim=0) * 0.18215
    x_enc = x_enc.reshape(B, T, 4, latent_sz, latent_sz)
    
    # Condition Variables
    x_cond = x_enc[:, :num_cond]
    x_cond = x_cond.unsqueeze(1).repeat(1, num_goals, 1, 1, 1, 1)
    x_cond = x_cond.reshape(B * num_goals, num_cond, 4, latent_sz, latent_sz).contiguous()
    
    z = torch.randn(B*num_goals, 4, latent_sz, latent_sz, device=dev).contiguous()
    
    # ---------------------------------------------------------
    # 2. CDiT DENOISING LOOP
    # ---------------------------------------------------------
    def trt_cdit_fn(x_in, t_in, y_in, x_cond_in, rel_t_in):
        inputs = {
            "x": x_in.contiguous().float(),
            "t": t_in.contiguous().float(),
            "y": y_in.contiguous().float(),
            "x_cond": x_cond_in.contiguous().float(),
            "rel_t": rel_t_in.contiguous().float()
        }
        
        out_shape = list(x_in.shape)
        out_shape[1] = 8 # Mean + Variance channels
        
        return run_trt_engine(cdit_context, inputs, "out", out_shape)

    model_kwargs = dict(y=y.contiguous().float(), x_cond=x_cond, rel_t=rel_t.contiguous().float())      
    
    samples = diff_obj.p_sample_loop(
        lambda x_val, t_val, **kw: trt_cdit_fn(x_val, t_val, kw["y"], kw["x_cond"], kw["rel_t"]), 
        z.shape, z, clip_denoised=False, model_kwargs=model_kwargs, progress=progress, device=dev
    )
    
    # ---------------------------------------------------------
    # 3. DECODE VAE
    # ---------------------------------------------------------
    samples_flat = samples.flatten(0, 1).contiguous().float()
    dec_list = []
    
    for i in range(samples_flat.shape[0]):
        z_in = samples_flat[i:i+1]
        px_out = run_trt_engine(dec_context, {"z": z_in}, "px", (1, 3, 224, 224))
        dec_list.append(px_out)
        
    decoded = torch.cat(dec_list, dim=0)
    
    return torch.clip(decoded, -1., 1.)

# Apply the global hook!
model_forward_wrapper = pure_trt_model_forward_wrapper
print("Successfully overrode `model_forward_wrapper` with UNIFIED Pure TensorRT pipeline!")
```

```python
def url_to_pil_image(url):
    response = requests.get(url)
    img = Image.open(BytesIO(response.content)).convert('RGB')
    return img

def load_internet_image(url):
    from torchvision import transforms
    _transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5], inplace=True),
    ])
    img = url_to_pil_image(url)
    x_start = _transform(img)
    return x_start.unsqueeze(0).expand(config['context_size'], x_start.shape[0], x_start.shape[1], x_start.shape[2])
    
# Jupyter Notebook Cell


# List of image links
# For unknown environments, please see https://github.com/facebookresearch/nwm/issues/7
image_links = [
    'https://raw.githubusercontent.com/amirbar/amirbar.github.io/refs/heads/master/images/recon.png',
    'https://raw.githubusercontent.com/amirbar/amirbar.github.io/refs/heads/master/images/scand.png',
    'https://raw.githubusercontent.com/amirbar/amirbar.github.io/refs/heads/master/images/sacson.png',
    'https://raw.githubusercontent.com/amirbar/amirbar.github.io/refs/heads/master/images/tartan.png'
]

# Output widget to hold the selected link
output = widgets.Output()
x_start_link = None  # This will hold the selected link

# Function to handle image click
def on_image_click(link):
    global x_start_link
    x_start_link = link
    with output:
        output.clear_output()
        print(f"Selected image link:\n{x_start_link}")

# Create HBox of images
image_buttons = []
for link in image_links:
    img = widgets.Button(
        description='click',
        layout=widgets.Layout(width='150px', height='20px', padding='0'),
        style={'button_color': 'lightgray'}
    )

    img._dom_classes += ('image-button',)
    img_link = link  # capture current link in closure

    def on_click(b, link=img_link):
        on_image_click(link)

    img.on_click(on_click)

    # Embed image using HTML style
    img_html = f'<img src="{link}" width="150px" height="150px">'
    img_html_widget = widgets.HTML(value=img_html)
    image_buttons.append(widgets.VBox([img_html_widget, img]))

# Display the gallery
display(widgets.HBox(image_buttons))
display(output)
```
